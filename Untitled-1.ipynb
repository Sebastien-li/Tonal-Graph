```python
from src.note_graph import NoteGraph, Music21Score
from pathlib import Path
import networkx as nx
import numpy as np
```

```python
file = Path(r'assets\scores\Mozart\Sonata K330\score_annotated.xml')
composer, title = file.parts[-3:-1]
score = Music21Score(file, composer, title)
note_graph = NoteGraph.from_m21score(score)
G_adj = note_graph.edge_index_to_adj()

file_subgraph = Path('subgraph_mozart.xml')
score_subgraph = Music21Score(file_subgraph, composer, title)
note_subgraph = NoteGraph.from_m21score(score_subgraph)
F_adj = note_subgraph.edge_index_to_adj()

G = nx.DiGraph(G_adj)
F = nx.DiGraph(F_adj)
```

```python
nx.algorithms.isomorphism.DiGraphMatcher(G,F).subgraph_is_isomorphic()
```

```python
nx.draw(G, with_labels=True)
```
